```python

from pint import UnitRegistry
ureg = UnitRegistry()

value = 123.456 * ureg.kilometer / ureg.hour
original = value

iterations = 200

for i in range(iterations):
    value = value.to(ureg.meter / ureg.second)
    value = value.to(ureg.mile / ureg.hour)
    value = value.to(ureg.foot / ureg.second)
    value = value.to(ureg.kilometer / ureg.hour)

difference = abs(original.magnitude - value.magnitude)

print("Original:", original)
print("Final:", value)
print("Difference:", difference)
```
