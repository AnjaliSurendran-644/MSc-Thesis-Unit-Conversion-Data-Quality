```python
import numpy as np
original_kmh = 123.456
KMH_TO_MPH = 0.621371192237334
MPH_TO_KMH = 1 / KMH_TO_MPH


speed = np.float32(original_kmh * KMH_TO_MPH)
iterations = 50
for i in range(iterations):
    speed = np.float32(speed * MPH_TO_KMH)
    speed = np.float32(speed * KMH_TO_MPH)


final_kmh = np.float32(speed * MPH_TO_KMH)


difference = abs(original_kmh - final_kmh)
percentage_error = (difference / original_kmh) * 100
print("Original:", original_kmh)
print("Final:", final_kmh)
print("Difference:", difference)
print("Percentage error:", percentage_error)
```
