```python
from pint import UnitRegistry

# Initialize unit registry
ureg = UnitRegistry()

# Initial value (km/h)
original_speed = 123.456 * ureg.kilometer / ureg.hour

# Copy for conversion loop
speed = original_speed

# Number of repeated conversions
iterations = 50

print(f"Original value: {original_speed}")

for i in range(iterations):
    # Convert km/h → mph
    speed = speed.to(ureg.mile / ureg.hour)

    # Convert mph → km/h
    speed = speed.to(ureg.kilometer / ureg.hour)

# Final value
print(f"Final value after {iterations} conversions: {speed}")

# Calculate percentage loss
difference = abs(original_speed.magnitude - speed.magnitude)
percentage_loss = (difference / original_speed.magnitude) * 100

print(f"Difference: {difference}")
print(f"Percentage loss: {percentage_loss:.10f}%")
```
