```python

# Real-world float-based simulation (correct error measurement)

original_kmh = 123.456  # ground truth in km/h

# precise conversion constants
KMH_TO_MPH = 0.621371192237334
MPH_TO_KMH = 1 / KMH_TO_MPH

# convert once to mph (simulate system storage)
speed = original_kmh * KMH_TO_MPH

iterations = 50

# repeated conversions
for i in range(iterations):
    speed = speed * MPH_TO_KMH
    speed = speed * KMH_TO_MPH

# convert final result back to km/h for comparison
final_kmh = speed * MPH_TO_KMH

# TRUE ERROR (compare to original km/h)
difference = abs(original_kmh - final_kmh)
percentage_error = (difference / original_kmh) * 100

print("Original (km/h):", original_kmh)
print("Final (km/h):", final_kmh)
print("Difference:", difference)
print("Percentage error:", percentage_error)
```
